## Exercise 2 — Get unique values with a set
Write a function `unique_values(nums)` that takes a list of numbers and **returns a set** of the unique values.

**Hint:** Converting a list to a set automatically removes duplicates.

```python
# TODO: implement unique_values
def unique_values(nums):
    numbers = set(nums)
    return numbers

my_nums=unique_values([1, 2, 2, 3, 3, 3, 4])
print(my_nums)
```

```python
# Tests — run me
sample = [1, 2, 2, 3, 3, 3, 4]
out = unique_values(sample)
assert isinstance(out, set), 'Return type should be a set.'
assert out == {1, 2, 3, 4}, 'Set should include each number once.'

# ensure result is not dependent on order
assert 3 in out and 5 not in out
print('✅ Exercise 2 passes!')
```

## Exercise 3 — Indexing & order
1. Implement `third_item(seq)` that returns the **item at index 2** from a list or tuple.
2. Implement `try_index_set(s)` that **tries** to access `s[0]`. If a `TypeError` occurs (because sets are unordered),
   return the string `'TypeError'`; otherwise return the accessed value.

**Goal:** Show that sets don’t support indexing.

```python
# TODO: implement the two functions
def third_item(seq):
    num_seq = list(seq)
    return num_seq[2]

sequence = ["Messi","Drogba","Kaka","Gatuso"]
item_three=third_item(sequence)
print(item_three)

def try_index_set(s):
    try:
      return s[0]
    except TypeError:
      return 'TypeError'
set_one = {"Messi","Drogba","Kaka","Gatuso"}
result = try_index_set(set_one)
print(result)
```

```python
# Tests — run me
lst = ['x', 'y', 'z', 'w']
tpl = ('x', 'y', 'z', 'w')
s = {'x', 'y', 'z', 'w'}

assert third_item(lst) == 'z'
assert third_item(tpl) == 'z'

res = try_index_set(s)
assert res == 'TypeError', 'Indexing a set should raise a TypeError.'
print('✅ Exercise 3 passes!')
```
